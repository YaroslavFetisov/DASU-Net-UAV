This notebook displays the high-resolution 300 DPI publication results:
1. **Reconstructed Fractional Abundance Maps** comparing Ground Truth vs Baseline ViT vs DASU-Net (Ours).
2. **Extracted Endmember Spectral Signatures** with dedicated sub-panels showing residual spectral errors $\Delta R(\lambda) = \hat{R}(\lambda) - R_{\text{GT}}(\lambda)$.


```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

# Robust path finder for runs or paper figures
def find_figure(fname):
    candidates = [
        Path('../runs/experiments_uav') / fname,
        Path('runs/experiments_uav') / fname,
        Path('../../paper') / fname,
        Path('../paper') / fname,
        Path('paper') / fname,
    ]
    for c in candidates:
        if c.exists():
            return c
    raise FileNotFoundError(f'Could not find {fname}')

# 1. Display Abundance Maps Comparison (Ground Truth vs Baseline vs DASU-Net)
path_abu = find_figure('abundance_comparison.png')
img_abu = Image.open(path_abu)
plt.figure(figsize=(12, 8), dpi=150)
plt.imshow(img_abu)
plt.axis('off')
plt.title('Figure 1: Reconstructed Abundance Maps Comparison (IEEE APUAVD 2026)', fontsize=12, fontweight='bold', pad=10)
plt.show()
```

```python
# 2. Display Endmember Spectral Signatures with Residual Error Sub-panels
path_spec = find_figure('spectral_signatures.png')
img_spec = Image.open(path_spec)
plt.figure(figsize=(12, 8), dpi=150)
plt.imshow(img_spec)
plt.axis('off')
plt.title('Figure 2: Extracted Endmember Reflectance & Residual Errors (IEEE APUAVD 2026)', fontsize=12, fontweight='bold', pad=10)
plt.show()
```
